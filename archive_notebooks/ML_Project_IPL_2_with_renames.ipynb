```python
df.loc[df['bowler']=="Brijesh Sharma",'bowler_cat'] = "RA_Pace"
df.loc[df["bowler"] == "Shivang Kumar", "bowler_cat"] = "Wrist_Spin"
df.loc[df['bowler']=="D Ferreira",'bowler_cat'] = "RA_Offspin"
df.loc[df['bowler']=="LPC Silva",'bowler_cat'] = "Wrist_Spin"
df.loc[df['bowler']=="AM Rahane",'bowler_cat'] = "RA_Offspin"
df.loc[df['bowler']=="FY Fazal",'bowler_cat'] = "RA_Pace"
df.loc[df['bowler']=="S Dhawan",'bowler_cat'] = "RA_Offspin"
df.loc[df['bowler']=="B Chipli",'bowler_cat'] = "RA_Pace"
df.loc[df['bowler']=="AC Gilchrist",'bowler_cat'] = "RA_Offspin"
df.loc[df['bowler']=="SS Agarwal",'bowler_cat'] = "Wrist_Spin"
df.loc[df['bowler']=="DA Warner",'bowler_cat'] = "Wrist_Spin"
df.loc[df['bowler']=="Ishan Kishan",'bowler_cat'] = "RA_Offspin"
df.loc[df['bowler']=="RS Ghosh",'bowler_cat'] = "RA_Pace"
df.loc[df['bowler']=="D Brevis",'bowler_cat'] = "Wrist_Spin"
df.loc[df['bowler']=="D Salunkhe",'bowler_cat'] = "Wrist_Spin"
df.loc[df['bowler']=="F du Plessis",'bowler_cat'] = "Wrist_Spin"
df.loc[df['bowler']=="P Ray Barman",'bowler_cat'] = "Wrist_Spin"
df.loc[df['bowler']=="DB Ravi Teja",'bowler_cat'] = "Wrist_Spin"
```

```python
df['bowler_cat'].value_counts()
```

```python
df_main = df
```

```python
matchup_conditions = [
    # Pace Same Arm
    (
        (df_main["batter_hand"] == "RHB") & (df_main["bowler_cat"] == "RA_Pace")
    )
    | (
        (df_main["batter_hand"] == "LHB") & (df_main["bowler_cat"] == "LA_Pace")
    ),
    # Pace Opposite Arm
    (
        (df_main["batter_hand"] == "RHB") & (df_main["bowler_cat"] == "LA_Pace")
    )
    | (
        (df_main["batter_hand"] == "LHB") & (df_main["bowler_cat"] == "RA_Pace")
    ),
    # Finger Spin Turning In
    (
        (df_main["batter_hand"] == "RHB")
        & (df_main["bowler_cat"] == "RA_Offspin")
    )
    | (
        (df_main["batter_hand"] == "LHB")
        & (df_main["bowler_cat"] == "LA_Orthodox")
    ),
    # Finger Spin Turning Away
    (
        (df_main["batter_hand"] == "RHB")
        & (df_main["bowler_cat"] == "LA_Orthodox")
    )
    | (
        (df_main["batter_hand"] == "LHB")
        & (df_main["bowler_cat"] == "RA_Offspin")
    ),
    # Wrist Spin (Leg-Spin & Chinaman dual-threat)
    (df_main["bowler_cat"] == "Wrist_Spin"),
]

matchup_choices = [
    "Pace_Same_Arm",
    "Pace_Opposite_Arm",
    "Finger_Spin_Turning_In",
    "Finger_Spin_Turning_Away",
    "Wrist_Spin",
]

df_main["matchup_type"] = np.select(
    matchup_conditions, matchup_choices, default="Unknown"
)

```

```python
df_main.to_csv('prog2.csv',index=False)
```
